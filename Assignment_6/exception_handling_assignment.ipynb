# Task 1: Safe Division Utility

```python
def safe_division():
    try:
        num = input("Enter numerator: ")
        den = input("Enter denominator: ")
        # Convert to float to handle decimals
        result = float(num) / float(den)
    except ValueError:
        print("Error: Input is not a number.")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")
    else:
        print(f"Result: {result}")
    finally:
        print("Operation Complete")
```

# Task 2: Bill Calculator with Error Handling

```python
def calculate_bill():
    prices = [120, 350, 'abc', 500, -200, 800]
    total = 0
    
    for p in prices:
        try:
            if not isinstance(p, (int, float)):
                raise TypeError
            if p < 0:
                raise ValueError("Negative price not allowed")
            total += p
        except TypeError:
            print(f"Skipping invalid item: {p}")
        except ValueError as e:
            print(f"Skipping: {e} ({p})")
            
    print(f"Running Total: {total}")
```

# Task 3: Custom Exception: Age Validator

```python
def check_age(age):
    if age < 1 or age > 120:
        raise ValueError("Age must be between 1 and 120")
    return age

def age_validator_tool():
    try:
        user_age = int(input("Enter your age: "))
        valid_age = check_age(user_age)
        print(f"Valid age recorded: {valid_age}")
    except ValueError as e:
        print(f"Custom Error: {e}")
```

# Task 4: File Reader with Exception Handling

```python
def safe_file_read():
    filename = input("Enter the filename to read: ")
    try:
        with open(filename, 'r') as f:
            lines = f.readlines()
            # Print first 3 lines safely
            for line in lines[:3]:
                print(line.strip())
    except FileNotFoundError:
        print("Error: The file does not exist.")
    except PermissionError:
        print("Error: You do not have permission to read this file.")
    finally:
        print("File operation attempted.")
```

# Task 5: Mini Program: Safe Shopping Cart

```python
def safe_shopping_cart():
    cart = []
    while True:
        choice = input("Enter product price (or 'q' to quit): ").lower()
        if choice == 'q':
            break
        try:
            price = float(choice)
            if price < 0:
                raise ValueError("Price cannot be negative.")
            cart.append(price)
        except ValueError as e:
            if "could not convert" in str(e):
                print("Invalid input. Please enter a numerical price.")
            else:
                print(f"Error: {e}")
                
    print(f"\nTotal items: {len(cart)}")
    print(f"Total bill: {sum(cart):.2f}")
```
